Real Code

```python
import numpy as np
import pandas as pd

names = ["Dhar", "Bhuvan", "Naman", "David", "Akash", 
         "Sujan", "Dawan", "Hannah", "Koushik", "Tapi"]

np.random.seed(42)
marks = np.random.randint(10, 100, size=(10, 10))

df = pd.DataFrame(marks, columns=[f"Sub{i+1}" for i in range(10)])
df.insert(0, "Name", names) 

print(df)

avg = np.mean(marks, axis=1)   
highest = np.max(marks, axis=1)   
lowest = np.min(marks, axis=1)

print("Name\t\tAverage\tHighest\tLowest")
for i, name in enumerate(names):
    print(f"{name:<10}\t{avg[i]:.2f}\t{highest[i]}\t{lowest[i]}")
```

```python
result = np.where(avg > 50, "Pass", "Fail")


df["Average"] = avg
df["Highest"] = highest
df["Lowest"] = lowest
df["Result"] = result


print(df)
```
