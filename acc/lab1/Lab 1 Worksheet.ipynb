## Question 1
Study the program. What is the program doing? How many passes (iterations of i loop) are executed?

Note: range(k) goes from 0 to k-1.

Answer: bubble sort, 3

```python
a=[5,1,4,2]
for i in range(3):
    for j in range(3-i):
        if a[j]>a[j+1]:
            a[j],a[j+1]=a[j+1],a[j]
print(a)
```

## Question 11

Write the code for optimized bubble sort using the idea in Question 7. Illustrate with an example.

```python
def optimized_sort(a):
    n=len(a)-1
    for i in range(n):
        flag=0
        for j in range(n-i):
            if a[j]>a[j+1]:
                a[j],a[j+1]=a[j+1],a[j]
                flag=1
        if flag==0:
            break
                
arr = [1,2,4,3]
optimized_sort(arr)
print(arr)
```
