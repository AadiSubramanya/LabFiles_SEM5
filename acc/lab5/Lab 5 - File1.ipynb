# Bucket Sort

This notebook implements **Bucket Sort** for numbers in the interval $[0,1]$.

Each bucket is sorted using **Insertion Sort**

## Main steps

1. Create $n$ empty buckets, where $n$ is the number of input values.
2. For each value $x \in [0,1]$, calculate the bucket index as $\lfloor nx \rfloor$.
3. If $x=1$, place it in the last bucket, whose index is $n-1$.
4. Sort every bucket using Insertion Sort.
5. Combine the sorted buckets to obtain the final sorted list.

## 1. Insertion Sort

The following function sorts one bucket in ascending order.

```python
def insertion_sort(bucket):
    # Start from the second element of the bucket
    for i in range(1, len(bucket)):
        # Store the current element
        key = bucket[i]

        # Begin comparison with the element immediately before it
        j = i - 1

        # Shift elements greater than key one position to the right
        while j >= 0 and bucket[j] > key:
            bucket[j + 1] = bucket[j]
            j = j - 1

        # Place key in its correct position
        bucket[j + 1] = key
```

## 2. Bucket Sort

This function accepts values only from the interval $[0,1]$. It displays the buckets before and after applying Insertion Sort.

```python
def bucket_sort(arr):
    # Find the number of input values
    n = len(arr)

    # Return an empty list if the input list is empty
    if n == 0:
        return []

    # Step 1: Create n empty buckets
    buckets = []
    for i in range(n):
        buckets.append([])

    # Step 2: Distribute the values among the buckets
    for value in arr:
        # Ensure that every value belongs to [0, 1]
        if value < 0 or value > 1:
            raise ValueError("All values must belong to the interval [0, 1].")

        # Calculate the appropriate bucket index
        index = int(n * value)

        # When value is 1, index becomes n.
        # Place it in the last bucket, whose index is n - 1.
        if index == n:
            index = n - 1

        # Add the value to the appropriate bucket
        buckets[index].append(value)

    print("Buckets before Insertion Sort:")
    for i in range(n):
        print("Bucket", i, ":", buckets[i])

    # Step 3: Sort each bucket using Insertion Sort
    for i in range(n):
        insertion_sort(buckets[i])

    print("\nBuckets after Insertion Sort:")
    for i in range(n):
        print("Bucket", i, ":", buckets[i])

    # Step 4: Combine all the sorted buckets
    sorted_arr = []
    for i in range(n):
        for value in buckets[i]:
            sorted_arr.append(value)

    return sorted_arr
```

## 3. Example

Run the following cell to see the distribution of values, the sorted buckets, and the final result.

```python
numbers = [
    0.78, 0.17, 0.39, 0.26, 0.72,
    0.94, 0.21, 0.12, 0.23, 0.68, 1.00
]

print("Original list:", numbers)
print()

result = bucket_sort(numbers)

print("\nFinal sorted list:", result)
```
