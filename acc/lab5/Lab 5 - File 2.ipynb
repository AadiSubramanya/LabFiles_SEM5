## Extend Bucket Sort to $[a,b]$

Complete the following program by replacing the **10 blanks**. The blanks are written as `____1____`, `____2____`, and so on.


```python
def bucket_sort_ab(A, a, b):
    n = len(A)

    # Width of each bucket
    width = (b - a) / n

    # Create n empty buckets
    buckets = [[] for i in range(n)]

    # Place each element in a suitable bucket
    for x in A:
        index = min(int(n * (x - a)/(b - a)), n - 1)
        buckets[index].append(x)

    # Display the interval and contents of each bucket
    for i in range(n):
        left = a + i * width
        right = left + width
        print("Bucket", i, "[", left, ",", right, "]:", buckets[i])

    # Sort each bucket
    for bucket in buckets:
        bucket.sort()

    # Combine the sorted buckets
    result = []
    for bucket in buckets:
        result.extend(bucket)

    return result
```

#### Question 1 Answer:
When value x=b, index becomes n (int part is n), but there is no bucket corresponding to index n (IndexError), therefore the bucket index is clamped to n-1 to make x go to the last bucket, preventing index error.


#### Question 2 Answer:

```python
#Functions

def insertion_sort(bucket):
    for i in range(1, len(bucket)):
        key = bucket[i]
        j = i - 1
        while j >= 0 and bucket[j] > key:
            bucket[j + 1] = bucket[j]
            j = j - 1
        bucket[j + 1] = key
```

```python
def bucket_sort_answer(A, a, b):
    n = len(A) #sets n=8 automatically
    width = (b - a) / n
    buckets = [[] for i in range(n)]
    for x in A:
        index = min(int(n * (x - a)/(b - a)), n - 1)
        buckets[index].append(x)
    for i in range(n):
        left = a + i * width
        right = left + width
        print("Bucket", i, "[", left, ",", right, "]:", buckets[i])
    for bucket in buckets:
        insertion_sort(bucket)
    result = []
    for bucket in buckets:
        result.extend(bucket)
    return result
```
