````python
#PART B
def recursive_max_path(rows, i=0, j=0):
    if i == len(rows) - 1:
        return rows[i][j], [rows[i][j]]
    left_sum, left_path = recursive_max_path(rows, i + 1, j)
    right_sum, right_path = recursive_max_path(rows, i + 1, j + 1)
    if left_sum > right_sum:
        return rows[i][j] + left_sum, [rows[i][j]] + left_path
    else:
        return rows[i][j] + right_sum, [rows[i][j]] + right_path

max_sum, path = recursive_max_path(rows)
print(f"Max Sum: {max_sum}")
print(f"Path: {path}")
````

````python
#PART D

def dp_max_path(rows):
    n = len(rows)
    dp = [row[:] for row in rows]
    for i in range(n - 2, -1, -1):
        for j in range(len(rows[i])):
            dp[i][j] += max(dp[i+1][j], dp[i+1][j+1])
    max_sum = dp[0][0]
    path = [rows[0][0]]
    curr_j = 0
    for i in range(n - 1):
        if dp[i+1][curr_j] > dp[i+1][curr_j+1]:
            pass
        else:
            curr_j += 1
        path.append(rows[i+1][curr_j])
    return max_sum, path

print("DP Result:", dp_max_path(rows))
````

### PART D

Recurrance Relation:

Base Case:$DP[n-1][j] = \text{rows}[n-1][j]$.

Relation:$DP[i][j] = \text{rows}[i][j] + \max(DP[i+1][j], DP[i+1][j+1])$

Time Complexity:$O(n^2)$

Space Complexity:$O(n^2)$

````python
#Part E
def pascal_triangle(n):
    triangle = []
    for i in range(n):
        row = [1] * (i + 1)
        for j in range(1, i):
            row[j] = triangle[i - 1][j - 1] + triangle[i - 1][j]
        triangle.append(row)
    return triangle


for n in [5, 10, 15, 20]:
    pt = pascal_triangle(n)
    max_sum, path = dp_max_path(pt)
    print(f"n = {n}")
    print(f"  Max sum: {max_sum}")
    print(f"  One max path: {path}")
    print()
````
