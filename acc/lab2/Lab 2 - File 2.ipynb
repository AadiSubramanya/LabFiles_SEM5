## Tower of Hanoi using three arrays (A, B, C).

- Array A starts with numbers 1..n, where index 0 is the TOP of the tower
  (i.e., A = [1, 2, 3, ..., n], so the smallest disk '1' is at the top).
- A "move" takes the top disk of the source array (pop(0)) and places it
  on top of the destination array (insert(0, disk)).
- The time taken by every individual pop(0) and insert(0, _) call is
  measured and accumulated, and the total is printed at the end.

```python
import time

# Global counters
total_time = 0.0
#move_count = 0


def timed_pop(peg):
    """Pop the top element (index 0) of a peg and record the time taken."""
    #start = time.perf_counter()
    #end = time.perf_counter()
    #total_time += (end - start)
    return peg.pop(0)


def timed_insert(peg, disk):
    """Insert a disk at the top (index 0) of a peg and record the time taken."""
    peg.insert(0, disk)



def move_disk(pegs, source, destination):
    """Move the top disk from `source` array to `destination` array."""
    #global total_time, move_count
    global total_time
    #disk = timed_pop(pegs[source])
    #timed_insert(pegs[destination], disk)
    start = time.perf_counter()
    timed_insert(pegs[destination], timed_pop(pegs[source]))
    end = time.perf_counter()
    total_time += (end - start)
    #move_count += 1
    #print(f"Move {move_count}: Disk {disk}  {source} -> {destination}")
    #print(f"   A={pegs['A']}  B={pegs['B']}  C={pegs['C']}")


def hanoi(n, pegs, source, auxiliary, destination):
    """
    Solve Tower of Hanoi recursively.

    Move n disks from `source` to `destination` using `auxiliary` as helper.
    """
    if n == 0:
        return
    # Step 1: Move n-1 disks from source to auxiliary, using destination as helper
    hanoi(n - 1, pegs, source, destination, auxiliary)

    # Step 2: Move the nth (largest remaining) disk from source to destination
    move_disk(pegs, source, destination)

    # Step 3: Move the n-1 disks from auxiliary to destination, using source as helper
    hanoi(n - 1, pegs, auxiliary, source, destination)



n = 4

#A holds disks 1..n with the smallest (1) at index 0 (the top)
pegs = {
    'A': list(range(1, n + 1)),
    'B': [],
    'C': []
}

print("\nInitial state:")
print(f"   A={pegs['A']}  B={pegs['B']}  C={pegs['C']}\n")

hanoi(n, pegs, 'A', 'B', 'C')

print("\nFinal state:")
print(f"   A={pegs['A']}  B={pegs['B']}  C={pegs['C']}")

#print(f"\nTotal moves: {move_count}")
print(f"\nTotal time spent in pop(0) and insert(0, _) operations: {total_time:.6f} seconds")
print(f"Time spent per move: {total_time/(2**n-1):.6f} seconds")

```

```python
def reverse_hanoi(n, pegs, source, auxiliary, destination):
    if n == 0:
        return
 
    # Move n-1 disks from source to auxiliary
    reverse_hanoi(n - 1, pegs, source, destination, auxiliary)
 
    # Move the nth disk from source to destination
    move_disk(pegs, source, destination)
 
    # Move n-1 disks from auxiliary to destination
    reverse_hanoi(n - 1, pegs, auxiliary, source, destination)

n = 31

pegs = {
    'A': list(range(n, 0, -1)),
    'B': [],
    'C': []
}

print("\nInitial state:")
print(f"   A={pegs['A']}  B={pegs['B']}  C={pegs['C']}\n")


reverse_hanoi(n, pegs, 'A', 'B', 'C')

print("\nFinal state:")
print(f"   A={pegs['A']}  B={pegs['B']}  C={pegs['C']}")
```
